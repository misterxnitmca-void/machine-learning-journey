```python
# import libraries
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
```

```python
#create a dataset
#this is independent variable  -- experience in years
x=np.array([
    [1],
    [2],
    [3],
    [4],
    [5]
])
# this is dependent variable
y=np.array([20,30,40,50,60])

```

```python
# create model
model = LinearRegression()
```

```python
#train your model
model.fit(x,y)
```

```python
prediction=model.predict([[6]])
print("expected salary of 6 year experience person is ",prediction)
```

```python
# print slope(m) and intercept (c)
print("slope:",model.coef_)
print("intercept",model.intercept_)
```

```python
# visual representation of our model is
plt.scatter(x, y, color="blue")
plt.plot(
    x,
    model.predict(x),
    color="red"
)
plt.xlabel("Experience")
plt.ylabel("Salary")
plt.title("Linear Regression")
plt.show()
```
